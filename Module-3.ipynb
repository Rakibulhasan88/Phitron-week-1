Conditional Statement in Python

```python
raining = True

if raining == True:
    print("It's raining. Don't forget your umbrella!")
    print("Stay dry and safe!")
else:
    print("It's not raining. Enjoy the sunshine!")
    print("Have a great day!")
```

Loop in Python

```python
for i in range(1, 11):
    print(i, "Hello World!")
```

```python
for i in range(1, 11):
    for j in range(1,5):
        print(i,j)
```

While Loop

```python
count = 1

while count < 10:
    print(count, "Hello world!")
    count += 1
    
print(count)
```

```python
sum = 0

for i in range(1, 11):
    
    print(i, "is processing...")
    if i % 2 == 0:
        continue
    
    sum += i
    print(i, "is added to sum.")
    
```
