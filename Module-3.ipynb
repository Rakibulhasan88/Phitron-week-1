While Loop

```python
count = 1

while count < 10:
    print(count, "Hello world!")
    count += 1
    
print(count)
```

```python
sum = 0

for i in range(1, 11):
    
    print(i, "is processing...")
    if i % 2 == 0:
        continue
    
    sum += i
    print(i, "is added to sum.")
    
```

Maximum & Minimum

```python
input_numbers = input()

numbers = input_numbers.split()

x = int(numbers[0])
y = int(numbers[1])
z = int(numbers[2])

minimum = x
maximum = x

#minimum finding
if y < minimum:
    minimum = y
if z < minimum:
    minimum = z

#maximum finding
if y > maximum:
    maximum = y
if z > maximum:
    maximum = z

print(minimum, maximum)
```
