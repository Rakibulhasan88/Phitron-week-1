```python
#immutable

lst = [10,20,30,40]

lst.append(50)
lst[1] = 100
print(lst)

tup = (10,20,30,40)
tup[1] = 100
print(tup)
```

```python
# empty structures
# Lists
B =[]
print(type(B), B)

# Tuples
C = ()
print(type(C), C)

# Dictionaries
# S = {}
# print(type(S), S)

# Sets
S = set()
print(type(S), S)
```

```python
S = {1,2,3,4,5}

if (3 in S):
    print("3 is present in the set")
else:
    print("3 is not present in the set")
    
sum = 0
    
for element in S:
    sum += element
    
print(sum)
```

```python
#mutable set

s={5,1,2,4,3}

s.add(6)
s.add(0)
print(s)

s.pop()
s.pop()
print(s)

s.remove(3)
print(s)
```

Set Mathematical Methods

```python
A={1,2,3,4,5}
B={1,2,3,4,5,6,7,8}

print(A.union(B))
print(A.intersection(B))

print(A.isdisjoint(B))

print(A.issubset(B))
```
