```python
tup = (10,20,30,40)
float_tup =(10.5, 10.2)

print(type(float_tup),float_tup)

mixed_tup = (10, 20.5, "Hello", True)

print(mixed_tup)

tup1 = (10,)

print(type(tup1), tup1)
```

```python
#accessing tuple elements

tup =(10,20,30,10.5,"str",False)
#indexing
print(tup[3])

#slicing
new_tup = tup[1:4]
print(new_tup)
```

```python
#immutable

lst = [10,20,30,40]

lst.append(50)
lst[1] = 100
print(lst)

tup = (10,20,30,40)
tup[1] = 100
print(tup)
```

```python
#tuple methods
tup = (10,20,30,40,10,20,30)
print(tup.count(10))
print(tup.index(30))
```

```python
S = {1,2,3,4,5}

if (3 in S):
    print("3 is present in the set")
else:
    print("3 is not present in the set")
    
sum = 0
    
for element in S:
    sum += element
    
print(sum)
```
